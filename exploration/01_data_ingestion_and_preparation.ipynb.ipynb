## Enriquecimento temporal da base

```python
df["date"] = df["datetime"].dt.normalize()
df["hour"] = df["datetime"].dt.hour
df["day_of_week"] = df["datetime"].dt.day_name()
df["day_of_week_num"] = df["datetime"].dt.dayofweek
df["month"] = df["datetime"].dt.month
df["year"] = df["datetime"].dt.year
df["is_weekend"] = df["day_of_week_num"] >= 5
```

```python
def get_season(month): # estações consideradas pelo hemisferio norte. são as estações metereologicas
    if month in [12, 1, 2]:
        return "winter"
    elif month in [3, 4, 5]:
        return "spring"
    elif month in [6, 7, 8]:
        return "summer"
    return "autumn"

df["season"] = df["month"].apply(get_season)
```

```python
output_path = Path("../exploration/data")
output_path.mkdir(parents=True, exist_ok=True)

df.to_parquet(output_path / "coffee_sales_prepared.parquet", index=False)
```

```python
sales_by_weekday = (
    df.groupby(["day_of_week_num", "day_of_week"])
      .agg(total_revenue=("money", "sum"))
      .reset_index()
      .sort_values("day_of_week_num")
)
```

```python
plt.figure(figsize=(10, 5))
plt.bar(sales_by_weekday["day_of_week"], sales_by_weekday["total_revenue"])
plt.title("Total Revenue by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
transactions_by_weekday = (
    df.groupby(["day_of_week_num", "day_of_week"])
      .agg(total_transactions=("money", "count"))
      .reset_index()
      .sort_values("day_of_week_num")
)
```

```python
plt.figure(figsize=(10, 5))
plt.bar(transactions_by_weekday["day_of_week"], transactions_by_weekday["total_transactions"])
plt.title("Total Transactions by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Total Transactions")
plt.xticks(rotation=45)
plt.show()
```
